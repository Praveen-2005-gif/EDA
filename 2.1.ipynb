```python
import numpy as np
import pandas as pd
data=pd.Series([45,67,89,23,56])
print(data)
print("values:",data.values)
print("Index:",data.index)
```

```python
population=pd.Series({'Mumbai':26548542,'Delhi':85469721,'Bengaluru':54967513,'Chennai':54697513,'Kolkata':65124879})
print(population)
print("Bengaluru:",population['Bengaluru'])
print(population['Delhi':'Chennai'])
```

```python
products={
    "Laptop":50000,"Phone":20000,"Headphone":3000 
}
prices=pd.Series(products)
print(prices)
print("Index matches dictionary keys:",prices.index.tolist() == list(products.keys()))
```

```python
area=pd.Series(
    [307421,458774,155229,451222,341254],
    index=["Rajasthan","Madhya pradesh","Karnataka","Gujarat","Tamil Nadu"]
)
population=pd.Series(
    [8.2,8.6,6.8,7.0,7.7],
    index=["Rajasthan","Madhya pradesh","Karnataka","Gujarat","Tamil Nadu"]
)
states=pd.DataFrame({
    "area": area,
    "population":population
})
states["density"]= states["population"]/states["area"]
print(states)
```

```python
indA=pd.Index([2,3,5,7,11])
indB=pd.Index([1,3,5,7,9])

print("Intersection:",indA.intersection(indB))
print("Union:",indA.union(indB))
print("Symmetric Differences:",indA.symmetric_difference(indB))
```
