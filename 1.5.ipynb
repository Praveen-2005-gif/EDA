```python
import numpy as np
books=np.array([
    ("Python basics","John Smith",450.0),
    ("Data Science","Alice Brown",550.0),
    ("Machine learning","David Lee",700.0),
    ("Numpy Guide","Robert John",350.0)
],dtype=[
    ("title","U30"),
    ("author","U20"),
    ("price","f8")
])
print(books)
```

```python
books=np.array([
    ("Python basics","John Smith",450.0),
    ("Data Science","Alice Brown",550.0),
    ("Machine learning","David Lee",700.0),
    ("Numpy Guide","Robert John",350.0)
],dtype=[
    ("title","U30"),
    ("author","U20"),
    ("price","f8")
])
# a
print(books["title"])
# b
print(books[1])
# c
print(books[books["price"]<500]["title"])
```

```python
# dictionary form
dtype_dict={
    "names":["title","author","price"],
    "formats":["U30","U20","f8"]
}
dtype1=np.dtype(dtype_dict)
print(dtype1)
# list of tuples form
dtype2=np.dtype([
    ("title","U30"),
    ("author","U20"),
    ("price","f8")
])
print(dtype2)
print(dtype1==dtype2)
    
```

```python
books=np.array([
    ("Python basics","John Smith",450.0),
    ("Data Science","Alice Brown",550.0),
    ("Machine learning","David Lee",700.0),
    ("Numpy Guide","Robert John",350.0)
],dtype=[
    ("title","U30"),
    ("author","U20"),
    ("price","f8")
])
data_rec=books.view(np.recarray)
print(data_rec.price)
```

```python
students=np.array([
    (101,"Alice",85.5),
    (102,"Bob",92.0),
    (103,"Charlie",78.5),
    (104,"David",95.0),
    (105,"Eve",88.0)
],dtype=[
    ("roll_no","i4"),
    ("name","U20"),
    ("marks","f8")
])
sort_students=np.sort(students,order="marks")[::-1]
print(sort_students)
```
